```python
import sys
sys.path.insert(0, '..')

import numpy as np
import matplotlib.pyplot as plt
from dotenv import load_dotenv
load_dotenv()
```

## 1. Définition du MDP PlateVision

```python
from modules.module_c.mdp_definition import get_mdp, STATES, ACTIONS

mdp = get_mdp()
print(f'États : {mdp["n_states"]} | Actions : {mdp["n_actions"]}')
print('\nÉtats définis :')
for s in STATES:
    print(f'  [{s.id}] {s.name} — {s.description}')
print('\nActions disponibles :')
for a in ACTIONS:
    print(f'  [{a.id}] {a.name} — {a.description}')
```

## 2. Visualisation de la matrice de récompenses R

```python
import seaborn as sns

R = mdp['R']
state_names = [s.name for s in STATES]
action_names = [a.name[:12] for a in ACTIONS]

plt.figure(figsize=(10, 6))
sns.heatmap(R, xticklabels=action_names, yticklabels=state_names,
            annot=True, fmt='.1f', cmap='RdYlGn', center=0)
plt.title('Matrice de Récompenses R[état, action] — MDP PlateVision')
plt.xlabel('Actions')
plt.ylabel('États')
plt.tight_layout()
plt.show()
```

## 3. Résolution par Value Iteration

```python
from modules.module_c.value_iteration import value_iteration

T, R = mdp['T'], mdp['R']
V_vi, policy_vi, history_vi = value_iteration(T, R, gamma=0.9)

print('Politique optimale (Value Iteration, γ=0.9) :')
for s, state in enumerate(STATES):
    action = ACTIONS[policy_vi[s]]
    print(f'  {state.name:20s} → {action.name:20s} | V*={V_vi[s]:.4f}')
```

## 4. Courbe de convergence VI

```python
plt.figure(figsize=(8, 4))
plt.semilogy(range(1, len(history_vi)+1), history_vi, 'b-o', markersize=4)
plt.axhline(y=1e-6, color='red', linestyle='--', label='Seuil θ=1e-6')
plt.xlabel('Itération')
plt.ylabel('Δ max (log)')
plt.title('Convergence Value Iteration — γ=0.9')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
print(f'Convergé en {len(history_vi)} itérations')
```
