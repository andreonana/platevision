```python
import os
from pathlib import Path

import cv2
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from dotenv import load_dotenv

load_dotenv()

DATA_RAW = Path(os.getenv('DATA_RAW_PATH', 'data/raw/'))
print(f'Dossier données brutes : {DATA_RAW.resolve()}')
```

## 1. Comptage et inventaire des images

```python
image_extensions = {'.jpg', '.jpeg', '.png', '.bmp'}
all_images = list(DATA_RAW.rglob('*'))
image_files = [f for f in all_images if f.suffix.lower() in image_extensions]

print(f'Nombre total d\'images : {len(image_files)}')
print(f'Extensions trouvées : {set(f.suffix for f in image_files)}')
```

## 2. Distribution des tailles d'images

```python
# À compléter : analyser les résolutions des images
widths, heights = [], []
for img_path in image_files[:200]:  # Limiter à 200 pour la rapidité
    img = cv2.imread(str(img_path))
    if img is not None:
        h, w = img.shape[:2]
        heights.append(h)
        widths.append(w)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].hist(widths, bins=30, color='steelblue', alpha=0.7)
axes[0].set_title('Distribution des largeurs')
axes[0].set_xlabel('Largeur (pixels)')

axes[1].hist(heights, bins=30, color='coral', alpha=0.7)
axes[1].set_title('Distribution des hauteurs')
axes[1].set_xlabel('Hauteur (pixels)')

plt.tight_layout()
plt.show()
print(f'Largeur moyenne : {np.mean(widths):.0f}px | Hauteur moyenne : {np.mean(heights):.0f}px')
```

## 3. Visualisation d'exemples du dataset

```python
# Afficher une grille d'exemples
sample_images = image_files[:16]
fig, axes = plt.subplots(4, 4, figsize=(14, 10))

for idx, (ax, img_path) in enumerate(zip(axes.flatten(), sample_images)):
    img = cv2.imread(str(img_path))
    if img is not None:
        img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        ax.imshow(img_rgb)
        ax.set_title(img_path.name[:15], fontsize=8)
    ax.axis('off')

plt.suptitle('Exemples du dataset PlateVision', fontsize=14)
plt.tight_layout()
plt.show()
```

## 4. Analyse des annotations YOLO

```python
DATA_ANNOTATIONS = Path(os.getenv('DATA_ANNOTATIONS_PATH', 'data/annotations/'))
annotation_files = list(DATA_ANNOTATIONS.glob('*.txt'))

print(f'Nombre de fichiers d\'annotation : {len(annotation_files)}')

# Distribution du nombre de plaques par image
plates_per_image = []
for ann_file in annotation_files:
    with open(ann_file) as f:
        lines = [l.strip() for l in f if l.strip()]
    plates_per_image.append(len(lines))

if plates_per_image:
    plt.figure(figsize=(8, 4))
    plt.hist(plates_per_image, bins=range(0, max(plates_per_image)+2), color='green', alpha=0.7)
    plt.title('Distribution du nombre de plaques par image')
    plt.xlabel('Nombre de plaques')
    plt.ylabel('Nombre d\'images')
    plt.show()
    print(f'Moyenne : {np.mean(plates_per_image):.2f} plaques/image')
```
